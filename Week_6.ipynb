Week-6: Pandas DataFrame Operations

```python
import pandas as pd

student_data = {
    "Student_ID": [201, 202, 203, 204],
    "Student_Name": ["Arjun", "Sneha", "Vikram", "Anjali"],
    "Score": [88, 76, 95, 69]
}

students = pd.DataFrame(student_data)

print("Student Data:")
print(students)
```

Using concat()

```python
import pandas as pd

batch1 = pd.DataFrame({
    "Student_Name": ["Arjun", "Sneha"],
    "Score": [88, 76]
})

batch2 = pd.DataFrame({
    "Student_Name": ["Vikram", "Anjali"],
    "Score": [95, 69]
})

combined_data = pd.concat([batch1, batch2], ignore_index=True)

print("Combined Student Data:")
print(combined_data)
```

c. Setting Conditions

```python
import pandas as pd

data = {
    "Student_Name": ["Arjun", "Sneha", "Vikram", "Anjali", "Rohan"],
    "Score": [88, 76, 95, 69, 72]
}

students = pd.DataFrame(data)

high_scorers = students[students["Score"] >= 75]

print("Students scoring 75 or above:")
print(high_scorers)
```

d. Adding a New Column

```python
import pandas as pd

data = {
    "Student_Name": ["Arjun", "Sneha", "Vikram", "Anjali"],
    "Score": [88, 76, 95, 39]
}

students = pd.DataFrame(data)

students["Result"] = students["Score"].apply(
    lambda x: "Pass" if x >= 40 else "Fail"
)

print("Student Result:")
print(students)
```
