This notebook follows the 3 guided stages:
1. **Load:** Ingest `campus_data.csv` and inspect the shape
2. **Clean & Aggregate :** Fill missing CGPA, department-wise averages, sorting
3. **Filter & Transform :** Filter top scorers and flag interview eligibility

## Setup: Run this cell first to create the sample dataset

```python
import pandas as pd
import numpy as np

data = {
    "StudentID": [101, 102, 103, 104, 105, 106],
    "Name": ["Aarav", "Bhavya", "Charan", "Divya", "Eshwar", "Farhan"],
    "Department": ["CSE", "ECE", "CSE", "IT", "ECE", "IT"],
    "CGPA": [8.2, 7.5, np.nan, 9.1, 6.8, np.nan],
    "CRT Score": [88, 62, 79, 95, 55, 84]
}
pd.DataFrame(data).to_csv("campus_data.csv", index=False)
```

## Stage 1 — Load
Ingest `campus_data.csv` using `pd.read_csv()` and print the dataset shape.

```python
df = pd.read_csv("campus_data.csv")
print("Dataset shape (rows, columns):", df.shape)
df
```

## Stage 2 — Clean & Aggregate
1. Fill missing CGPA values with the column average using `.fillna()`
2. Calculate the average CodingScore department-wise using `.groupby()`
3. Sort students by CodingScore from highest to lowest

```python
# 2.1 Fill missing CGPA with the column average
df["CGPA"] = df["CGPA"].fillna(df["CGPA"].mean())
print("Missing CGPA values remaining:", df["CGPA"].isna().sum())
df
```

```python
# 2.2 Average CodingScore department-wise
dept_avg = df.groupby("Department")["CRT Score"].mean()
print("Department-wise average CodingScore:")
dept_avg
```

```python
# 2.3 Sort by CodingScore, highest to lowest
df_sorted = df.sort_values(by="CRT Score", ascending=False).reset_index(drop=True)
df_sorted
```

## Stage 3 — Filter & Transform
1. Filter students with CodingScore >= 80
2. Add a new column `Interview_Eligible` set to `"Yes"` for all qualifying students

```python
# 3.1 Filter students with CodingScore >= 80
eligible = df[df["CRT Score"] >= 80].copy()

# 3.2 Add the Interview_Eligible column
eligible["Interview_Eligible"] = "Yes"

print("Number of interview-eligible students:", len(eligible))
eligible
```
